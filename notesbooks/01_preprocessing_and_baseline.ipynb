```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/Titanic-Dataset.csv")
```

```python
# Selecting the target column
X = df.drop(columns = ["Survived"])
y = df["Survived"]
```

```python
X.head()
```

```python
# Identifying the probelm type
problem_type = None
if y.dtype == 'object':
    problem_type = 'classification'
elif y.nunique() <= 10:
    problem_type = 'classification'
else:
    problem_type = 'regression'

print(problem_type)
```

```python
# train-test-split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = 10)
```

```python
# Identifying the feature type
numerical_feature = X.select_dtypes(include = 'number').columns
categorical_feature = X.select_dtypes(include = 'object').columns
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
# Imputing missing values and scaling for numerical features
numeric_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'median')),
    ('scaling',StandardScaler())
])
```

```python
# Imputing missing values and scaling for categorical features
categorical_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('encoder',OneHotEncoder(handle_unknown = 'ignore'))
    
])
```

```python
# Columntransforming
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer([
    ('numeric_pipe',numeric_pipeline,numerical_feature),
    ('categoric_pipe',categorical_pipeline,categorical_feature)
])
```

```python
# Training 1st regression model
from sklearn.linear_model import LinearRegression
model = Pipeline([
    ('preprosser',transformer),
    ('model',LinearRegression())
])
model.fit(X_train,y_train)

```

```python
y_pred = model.predict(X_test)
```

```python
# 1st Model scores
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("r2_score: ",r2)
```

```python
# Training 2nd regression model
from sklearn.tree import DecisionTreeRegressor

model = Pipeline([
    ('preprosser',transformer),
    ('model',DecisionTreeRegressor())
])

model.fit(X_train,y_train)
    

```

```python
y_pred = model.predict(X_test)
```

```python
# 2nd model scores
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("r2_score: ",r2)
```

```python
# Training 3rd regression model
from sklearn.ensemble import RandomForestRegressor
model = Pipeline([
    ('preprosser',transformer),
    ('model',RandomForestRegressor())
])

model.fit(X_train,y_train)
    
```

```python
y_pred = model.predict(X_test)
```

```python
# 3rd model socres
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("r2_score: ",r2)
```
