```python
health_summary = {
    "rows": len(df),
    "columns": len(df.columns),
    "duplicate_rows": df.duplicated().sum(),
    "missing_columns": (df.isnull().sum() > 0).sum(),
    "constant_columns": len(constant_cols),
    "id_like_columns": len(id_like_cols)
}

health_summary
```

```python
warnings = []

if health_summary["duplicate_rows"] > 0:
    warnings.append(f"⚠️ {health_summary['duplicate_rows']} duplicate rows found.")

if health_summary["missing_columns"] > 0:
    warnings.append(f"⚠️ {health_summary['missing_columns']} columns contain missing values.")

if health_summary["constant_columns"] > 0:
    warnings.append(f"⚠️ {health_summary['constant_columns']} constant columns found.")

if health_summary["id_like_columns"] > 0:
    warnings.append(f"⚠️ {health_summary['id_like_columns']} possible ID-like columns found.")

warnings
```

```python
for col in df.columns:
    pct = missing_percent[col]

    if pct >= 50:
        warnings.append(f"🔴 {col} has {pct:.1f}% missing values.")
    elif pct > 0:
        warnings.append(f"🟡 {col} has {pct:.1f}% missing values.")
```

```python
print("DATASET HEALTH CHECK")
print("-" * 30)

for key, value in health_summary.items():
    print(f"{key}: {value}")

print("\nWARNINGS")
for warning in warnings:
    print(warning)
```

```python
problematic_cols = {
    "missing": list(df.columns[df.isnull().any()]),
    "constant": list(constant_cols.index),
    "id_like": list(id_like_cols.index)
}

problematic_cols
```

```python
if len(warnings) == 0:
    health_status = "Healthy"
elif len(warnings) <= 3:
    health_status = "Needs Attention"
else:
    health_status = "Poor"

health_status
```

```python
target_distribution = df["Survived"].value_counts(normalize=True)
minority_pct = target_distribution.min() * 100

if minority_pct < 20:
    warnings.append(
        f"🔴 Target class imbalance detected: minority class is only {minority_pct:.1f}%."
    )
elif minority_pct < 30:
    warnings.append(
        f"🟡 Target class imbalance detected: minority class is {minority_pct:.1f}%."
    )
```

```python
if len(warnings) == 0:
    health_status = "Healthy"
elif len(warnings) <= 3:
    health_status = "Needs Attention"
else:
    health_status = "Poor"

print("Final Health Status:", health_status)
print("\nWarnings:")
for warning in warnings:
    print(warning)
```
