```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

```

```python
df = pd.read_csv("../data/Titanic-Dataset.csv")
```

```python
# Selecting the target column
X = df.drop(columns = ["Survived"])
y = df["Survived"]
```

```python
# Identifying the feature type
numerical_feature = X.select_dtypes(include = 'number').columns
categorical_feature = X.select_dtypes(include = 'object').columns
```

```python
# Imputing missing values and scaling for numerical features
numeric_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'median')),
    ('scaling',StandardScaler())
])
```

```python
# Imputing missing values and scaling for categorical features
categorical_pipeline = Pipeline([
    ('imputer',SimpleImputer(strategy = 'most_frequent')),
    ('encoder',OneHotEncoder(handle_unknown = 'ignore'))
    
])
```

```python
# Columntransforming
transformer = ColumnTransformer([
    ('numeric_pipe',numeric_pipeline,numerical_feature),
    ('categoric_pipe',categorical_pipeline,categorical_feature)
])
```

```python
from sklearn.linear_model import LinearRegression
lr_pipeline = Pipeline([
    ('preprosser',transformer),
    ('model',LinearRegression())
])

```

```python
from sklearn.model_selection import cross_val_score

# performing cross_validation on LinearRegression
scores = cross_val_score(
    lr_pipeline,
    X,
    y,
    cv=5,
    scoring = 'r2'
)

```

```python
print(scores)
print(scores.mean())
scores.std()
```

```python
from sklearn.tree import DecisionTreeRegressor


dtr_pipeline = Pipeline([
    ('preprosser',transformer),
    ('model',DecisionTreeRegressor())
])
```

```python
# performing cross_validation on DecisionTreeRefressor
scores = cross_val_score(
    dtr_pipeline,
    X,
    y,
    cv=5,
    scoring = 'r2'
)

```

```python
print(scores)
print(scores.mean())
scores.std()
```

```python
from sklearn.ensemble import RandomForestRegressor
Rfr_pipeline = Pipeline([
    ('preprosser',transformer),
    ('model',RandomForestRegressor())
])


 
```

```python
# performing cross_validation on RandomForestRegressor
scores = cross_val_score(
    Rfr_pipeline,
    X,
    y,
    cv=5,
    scoring = 'r2'
)
```

```python
print(scores)
print(scores.mean())
scores.std()
```
